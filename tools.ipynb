```python
from anthropic import Anthropic
from dotenv import load_dotenv
from datetime import datetime

load_dotenv()
client = Anthropic()

model = "claude-haiku-4-5"
msg_queue = []
```

### Chat Functions

```python
def add_msg(msg_queue, msg, role):
    msg_queue.append({
        "content": msg,
        "role": role
    })

def claude_msg(msg_queue, extra_args=None, stream=False):
    params = {
        "model": model,
        "max_tokens": 1000,
        "extra_headers": {"anthropic-workspace-id": "wrkspc_01RNMpstasBQ8NVXbRqZoota"},
        "messages": msg_queue
    }

    if extra_args:
        params.update(extra_args)

    if stream:
        return client.messages.stream(**params)
    else:
        return client.messages.create(**params)

def chat(msg_queue, msg, extra_args=None, prefill_message=None):
    add_msg(msg_queue, msg,"user")
    if prefill_message: add_msg(msg_queue, prefill_message, "assistant")

    return claude_msg(msg_queue, extra_args, True)
```

```python
def get_current_time(date_format="%d/%m/%Y, %H:%M:%S"):
    return datetime.now().strftime(format = date_format)

def set_reminder_time(reminder_duration):
    if (not reminder_duration):
        raise ValueError("Error: reminder_duration Not Supplied.")

    get_current_time("")
    if reminder_duration: return datetime()
```
